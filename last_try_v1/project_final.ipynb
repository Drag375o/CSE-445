#Phase 3 : 1st cell (drop redundant Kelvin columns, keep Celsius)

```python
# ============================================================
# PHASE 3 — MULTI-CITY AGGREGATION & MERGING
# CELL 1 — Drop redundant Kelvin temperature columns
# ============================================================

# We already have temp_c, temp_min_c, temp_max_c — keeping the Kelvin
# originals alongside them is redundant and risks accidentally feeding
# both unit systems into the model later.
weather = weather.drop(columns=["temp", "temp_min", "temp_max"])
weather = weather.rename(columns={
    "temp_c": "temp",
    "temp_min_c": "temp_min",
    "temp_max_c": "temp_max"
})

print("Weather columns now:", weather.columns.tolist())
```

```python
# ============================================================
# PHASE 3 — MULTI-CITY AGGREGATION & MERGING
# CELL 2 — Aggregate 5-city weather into a single Spain-wide series
# ============================================================

# Approximate 2018 municipal populations (Spain's INE register).
# Used as weights so that more populous cities (which represent more
# actual electricity consumers) influence the "Spain-wide" weather
# signal more than smaller cities.
city_populations = {
    "Madrid": 3_223_000,
    "Barcelona": 1_620_000,
    "Valencia": 791_000,
    "Seville": 689_000,
    "Bilbao": 345_000
}

weather["pop_weight"] = weather["city_name"].map(city_populations)

numeric_cols = ["temp", "temp_min", "temp_max", "pressure", "humidity",
                 "wind_speed", "wind_deg", "rain_1h", "rain_3h",
                 "snow_3h", "clouds_all"]

def weighted_avg(group):
    weights = group["pop_weight"]
    return pd.Series({
        col: (group[col] * weights).sum() / weights.sum()
        for col in numeric_cols
    })

weather_spain = (
    weather.groupby("datetime_utc")
    .apply(weighted_avg)
    .reset_index()
)

print("Spain-wide weather shape:", weather_spain.shape)
print(weather_spain.head())
```

```python
# ============================================================
# PHASE 3 — MULTI-CITY AGGREGATION & MERGING
# CELL 3 — Merge Spain-wide weather with electricity data
# ============================================================

print("Electricity rows before merge:", len(electricity))
print("Weather rows before merge:", len(weather_spain))

merged = pd.merge(
    electricity,
    weather_spain,
    on="datetime_utc",
    how="inner"
)

print("\nMerged shape:", merged.shape)

# Check for any timestamps that didn't find a match on either side
elec_only = set(electricity["datetime_utc"]) - set(weather_spain["datetime_utc"])
weather_only = set(weather_spain["datetime_utc"]) - set(electricity["datetime_utc"])
print("\nTimestamps in electricity but not weather:", len(elec_only))
print("Timestamps in weather but not electricity:", len(weather_only))

# Check the merge didn't duplicate any rows
print("\nDuplicate datetime_utc rows after merge:",
      merged.duplicated(subset=["datetime_utc"]).sum())

# Check for any missing values introduced by the merge
print("\nMissing values introduced by merge:")
merge_missing = merged.isna().sum()
print(merge_missing[merge_missing > 0] if merge_missing.sum() > 0 else "None")

print("\nMerged columns:", merged.columns.tolist())
```

#Phase 4 : Feature Engineering

```python
# ============================================================
# PHASE 4 — FEATURE ENGINEERING
# CELL 1 — Time-based features (using Spain local time, not UTC)
# ============================================================

# Convert to Europe/Madrid local time ONLY for extracting calendar features —
# demand patterns (rush hour, weekday routines) follow local clock time,
# and using raw UTC would misalign these features across DST changes.
local_time = merged["datetime_utc"].dt.tz_convert("Europe/Madrid")

merged["hour"] = local_time.dt.hour
merged["day_of_week"] = local_time.dt.dayofweek       # 0=Monday, 6=Sunday
merged["month"] = local_time.dt.month
merged["is_weekend"] = (merged["day_of_week"] >= 5).astype(int)

# Season, based on meteorological seasons (Northern Hemisphere)
def month_to_season(month):
    if month in [12, 1, 2]:
        return "winter"
    elif month in [3, 4, 5]:
        return "spring"
    elif month in [6, 7, 8]:
        return "summer"
    else:
        return "autumn"

merged["season"] = merged["month"].apply(month_to_season)

print(merged[["datetime_utc", "hour", "day_of_week", "month", "is_weekend", "season"]].head(10))
print("\nSeason value counts:")
print(merged["season"].value_counts())
```

```python
# ============================================================
# PHASE 4 — FEATURE ENGINEERING
# CELL 2 — Lagged & rolling demand features (no leakage)
# ============================================================

# Sort chronologically first — lag features are meaningless (and dangerous)
# on unsorted data.
merged = merged.sort_values("datetime_utc").reset_index(drop=True)

target = "total load actual"

# Lag features: demand at t-1 hour, t-24 hours (same hour yesterday),
# and t-168 hours (same hour, same weekday, one week ago).
# Each of these uses ONLY information available before time t — no leakage.
merged["demand_lag_1h"] = merged[target].shift(1)
merged["demand_lag_24h"] = merged[target].shift(24)
merged["demand_lag_168h"] = merged[target].shift(168)

# Rolling statistics: mean/std of the PRECEDING 24 hours, not including
# the current hour itself. shift(1) before rolling ensures the window
# ends at t-1, not t.
merged["demand_rolling_mean_24h"] = merged[target].shift(1).rolling(window=24).mean()
merged["demand_rolling_std_24h"] = merged[target].shift(1).rolling(window=24).std()

print("New columns:")
print(merged[["datetime_utc", target, "demand_lag_1h", "demand_lag_24h",
              "demand_lag_168h", "demand_rolling_mean_24h", "demand_rolling_std_24h"]].head(10))

print("\nMissing values from lag features (expected at the start of the series):")
lag_cols = ["demand_lag_1h", "demand_lag_24h", "demand_lag_168h",
            "demand_rolling_mean_24h", "demand_rolling_std_24h"]
print(merged[lag_cols].isna().sum())
```

```python
# ============================================================
# PHASE 4 — FEATURE ENGINEERING
# CELL 3 — Finalize feature set for regression modeling
# ============================================================

# These are the columns we'll actually use as MODEL INPUTS for regression.
# Note: 'merged' still keeps all the generation/price columns too — we're
# not deleting them, just defining which ones feed the model. The
# generation-source columns stay in 'merged' because we'll need them
# later for the CO2 analysis (Phase 13), even though they're not
# predictors of demand.

feature_cols = [
    # weather (same-hour only)
    "temp", "temp_min", "temp_max", "humidity", "wind_speed",
    "pressure", "rain_1h", "clouds_all",
    # time
    "hour", "day_of_week", "month", "is_weekend", "season",
    # demand history
    "demand_lag_1h", "demand_lag_24h", "demand_lag_168h",
    "demand_rolling_mean_24h", "demand_rolling_std_24h"
]

target_col = "total load actual"

print("Feature columns:", feature_cols)
print("\nNumber of features:", len(feature_cols))
print("\nMissing values in feature set:")
print(merged[feature_cols].isna().sum())

print("\nData types:")
print(merged[feature_cols].dtypes)
```

#Phase 7 : Regression Models

```python
# ============================================================
# PHASE 7 — REGRESSION MODELS
# CELL 1 — Prepare features (X) and target (y) for modeling
# ============================================================

# Dropping temp_min/temp_max: Phase 5's heatmap showed they're
# near-perfectly correlated with temp (0.96-0.99), so they add
# redundancy/multicollinearity without new information.
model_features = [
    "temp", "humidity", "wind_speed", "pressure", "rain_1h", "clouds_all",
    "hour", "day_of_week", "month", "is_weekend", "season",
    "demand_lag_1h", "demand_lag_24h", "demand_lag_168h",
    "demand_rolling_mean_24h", "demand_rolling_std_24h"
]

target_col = "total load actual"

# One-hot encode 'season' (the only non-numeric feature). We fit the
# encoding categories on TRAIN only, then apply the same columns to
# test, so test can't introduce categories the model never saw.
train_X = pd.get_dummies(train_df[model_features], columns=["season"], drop_first=True)
test_X = pd.get_dummies(test_df[model_features], columns=["season"], drop_first=True)

# Align test columns to train columns (in case a category is missing in test)
test_X = test_X.reindex(columns=train_X.columns, fill_value=0)

train_y = train_df[target_col]
test_y = test_df[target_col]

print("Train X shape:", train_X.shape)
print("Test X shape:", test_X.shape)
print("\nFeature columns:", train_X.columns.tolist())
```

```python
# ============================================================
# PHASE 7 — REGRESSION MODELS
# CELL 2 — Baseline model: naive persistence (predict = demand_lag_1h)
# ============================================================

import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

baseline_pred = test_df["demand_lag_1h"]

baseline_mae = mean_absolute_error(test_y, baseline_pred)
baseline_rmse = np.sqrt(mean_squared_error(test_y, baseline_pred))
baseline_r2 = r2_score(test_y, baseline_pred)

print("Baseline (naive persistence) performance on test set:")
print(f"MAE:  {baseline_mae:.2f} MW")
print(f"RMSE: {baseline_rmse:.2f} MW")
print(f"R²:   {baseline_r2:.4f}")
```

```python
# ============================================================
# PHASE 7 — REGRESSION MODELS
# CELL 3 — Train Linear Regression, Random Forest, Gradient Boosting
# ============================================================

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

# --- Linear Regression ---
# What it does: fits a straight-line (hyperplane) relationship between
# features and demand. Fast, fully interpretable coefficients.
# Why we expect it to underperform: Phase 5 showed temperature has a
# U-shaped, non-linear relationship with demand — a straight line
# structurally cannot represent that curve.
lr_model = LinearRegression()
lr_model.fit(train_X, train_y)

# --- Random Forest Regressor ---
# What it does: builds many decision trees on random subsets of data/features,
# averages their predictions. Naturally captures non-linear relationships
# (like the U-shape) and feature interactions (e.g. temp effect depends on hour).
# Key hyperparameters: n_estimators (number of trees), max_depth (tree size,
# controls overfitting).
rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(train_X, train_y)

# --- Gradient Boosting Regressor ---
# What it does: builds trees sequentially, where each new tree corrects the
# errors of the previous ones. Usually the strongest of these three, but
# slower to train and more sensitive to hyperparameters (risk of overfitting
# if too many estimators / too high learning rate).
gb_model = GradientBoostingRegressor(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.1,
    random_state=42
)
gb_model.fit(train_X, train_y)

print("All three models trained.")
```

```python
# ============================================================
# PHASE 7 — REGRESSION MODELS
# CELL 4 — Evaluate all models on the test set, build comparison table
# ============================================================

models = {
    "Linear Regression": lr_model,
    "Random Forest": rf_model,
    "Gradient Boosting": gb_model
}

results = []

# Include the baseline from Cell 2 for direct comparison
results.append({
    "Model": "Baseline (naive persistence)",
    "MAE": baseline_mae,
    "MSE": baseline_rmse ** 2,
    "RMSE": baseline_rmse,
    "R2": baseline_r2
})

predictions = {}  # save for plotting later

for name, model in models.items():
    pred = model.predict(test_X)
    predictions[name] = pred

    mae = mean_absolute_error(test_y, pred)
    mse = mean_squared_error(test_y, pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(test_y, pred)

    results.append({"Model": name, "MAE": mae, "MSE": mse, "RMSE": rmse, "R2": r2})

results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))
```

```python
# ============================================================
# PHASE 8 — TEMPERATURE RESPONSE MODEL
# CELL 1 — Feature builder (weather + calendar only, no demand lags)
# ============================================================

import numpy as np
import pandas as pd

# Balance-point temperature: below this, people heat; above it, they cool.
# 18 °C is the international convention for degree-day calculations.
# We test alternatives in Cell 8.10.
BASE_TEMP = 18.0

# Weather and calendar columns that pass through unchanged.
PASSTHROUGH = ["humidity", "wind_speed", "pressure", "rain_1h", "clouds_all",
               "hour", "day_of_week", "month", "is_weekend"]

# Fixed category list so train and test always produce identical dummy columns.
SEASON_CATEGORIES = ["autumn", "spring", "summer", "winter"]


def build_response_features(df, temp_shift=0.0, ref_columns=None,
                            drop_calendar_season=False):
    """Build the Model B design matrix.

    temp_shift: degrees C added to temperature. CRITICAL — cdd and hdd are
    RECOMPUTED from the shifted temperature. If we shifted `temp` but left
    cdd/hdd at their original values, the feature matrix would be internally
    inconsistent and the scenario would be meaningless.
    """
    shifted = df["temp"] + temp_shift

    X = pd.DataFrame(index=df.index)
    X["temp"] = shifted
    X["cdd"] = (shifted - BASE_TEMP).clip(lower=0)   # cooling degrees
    X["hdd"] = (BASE_TEMP - shifted).clip(lower=0)   # heating degrees

    for col in PASSTHROUGH:
        X[col] = df[col]

    # One-hot encode season with a fixed category list (drop_first avoids
    # the dummy-variable trap; 'autumn' becomes the reference category).
    season = pd.Categorical(df["season"], categories=SEASON_CATEGORIES)
    dummies = pd.get_dummies(season, prefix="season", drop_first=True).astype(int)
    dummies.index = X.index
    X = pd.concat([X, dummies], axis=1)

    # Used only for the Cell 8.10 sensitivity check.
    if drop_calendar_season:
        X = X.drop(columns=[c for c in X.columns
                            if c == "month" or c.startswith("season_")])

    # Force test columns into the exact same order as train.
    if ref_columns is not None:
        X = X.reindex(columns=ref_columns, fill_value=0)

    return X


train_Xr = build_response_features(train_df)
test_Xr  = build_response_features(test_df, ref_columns=train_Xr.columns)

train_yr = train_df["total load actual"]
test_yr  = test_df["total load actual"]

print("Model B train matrix:", train_Xr.shape)
print("Model B test matrix: ", test_Xr.shape)
print("\nFeatures:", train_Xr.columns.tolist())
print("\nColumns identical between train and test:",
      list(train_Xr.columns) == list(test_Xr.columns))
print("\nNo demand-lag columns present:",
      not any("demand" in c for c in train_Xr.columns))
```

```python
# ============================================================
# PHASE 8 — CELL 2 — Train the response model, compare to Model A
# ============================================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# min_samples_leaf=5 forces each leaf to average over at least 5 hours.
# This matters here: with leaves of size 1 the model memorises individual
# hours, and predictions become jumpy step functions — which is exactly the
# numerical noise that polluted your old scenario results.
rf_response = RandomForestRegressor(
    n_estimators=300,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)
rf_response.fit(train_Xr, train_yr)

pred_response = rf_response.predict(test_Xr)

comparison = pd.DataFrame([
    {
        "Model": "A: forecasting (with demand lags)",
        "MAE": mean_absolute_error(test_y, predictions["Random Forest"]),
        "RMSE": np.sqrt(mean_squared_error(test_y, predictions["Random Forest"])),
        "R2": r2_score(test_y, predictions["Random Forest"]),
        "Valid for counterfactuals?": "NO"
    },
    {
        "Model": "B: response (weather + calendar only)",
        "MAE": mean_absolute_error(test_yr, pred_response),
        "RMSE": np.sqrt(mean_squared_error(test_yr, pred_response)),
        "R2": r2_score(test_yr, pred_response),
        "Valid for counterfactuals?": "YES"
    },
])

print(comparison.round(3).to_string(index=False))
```

```python
# ============================================================
# PHASE 8 — CELL 3 — Linear HDD/CDD model (interpretable coefficients)
# ============================================================

from sklearn.linear_model import LinearRegression


def build_linear_features(df, temp_shift=0.0, ref_columns=None):
    """Design matrix for the linear model.

    Hour, day-of-week and month are ONE-HOT encoded here rather than left as
    integers. A linear model would otherwise read hour=23 as '23 times more
    of something' than hour=1, which is nonsense. Trees don't care; linear
    models very much do.
    """
    shifted = df["temp"] + temp_shift

    X = pd.DataFrame(index=df.index)
    X["cdd"] = (shifted - BASE_TEMP).clip(lower=0)
    X["hdd"] = (BASE_TEMP - shifted).clip(lower=0)
    X["humidity"] = df["humidity"]
    X["wind_speed"] = df["wind_speed"]

    for col, prefix in [("hour", "h"), ("day_of_week", "dow"), ("month", "m")]:
        d = pd.get_dummies(df[col].astype(int), prefix=prefix,
                           drop_first=True).astype(int)
        X = pd.concat([X, d], axis=1)

    if ref_columns is not None:
        X = X.reindex(columns=ref_columns, fill_value=0)

    return X


train_Xl = build_linear_features(train_df)
test_Xl  = build_linear_features(test_df, ref_columns=train_Xl.columns)

lin_response = LinearRegression().fit(train_Xl, train_yr)
pred_lin = lin_response.predict(test_Xl)

coefs = pd.Series(lin_response.coef_, index=train_Xl.columns)

print("Linear response model performance:")
print(f"  MAE:  {mean_absolute_error(test_yr, pred_lin):.1f} MW")
print(f"  RMSE: {np.sqrt(mean_squared_error(test_yr, pred_lin)):.1f} MW")
print(f"  R2:   {r2_score(test_yr, pred_lin):.4f}")

print("\n--- TEMPERATURE SENSITIVITY (the headline numbers) ---")
print(f"  Cooling: {coefs['cdd']:+.1f} MW per degree C above {BASE_TEMP:.0f} C")
print(f"  Heating: {coefs['hdd']:+.1f} MW per degree C below {BASE_TEMP:.0f} C")
```

```python
# ============================================================
# PHASE 8 — CELL 4 — Permutation importance for Model B
# ============================================================

from sklearn.inspection import permutation_importance

# Permutation importance shuffles one column at a time and measures how much
# the model gets WORSE. Unlike Random Forest's built-in importance (which is
# computed on training data and biased toward high-cardinality features),
# this is measured on the TEST set, so it reflects genuine predictive value.
perm = permutation_importance(
    rf_response, test_Xr, test_yr,
    n_repeats=10,
    random_state=42,
    n_jobs=-1,
    scoring="neg_mean_absolute_error"
)

perm_importance = pd.Series(
    perm.importances_mean, index=test_Xr.columns
).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(9, 7))
ax.barh(perm_importance.index, perm_importance.values,
        xerr=pd.Series(perm.importances_std, index=test_Xr.columns)[perm_importance.index])
ax.set_xlabel("Increase in MAE when feature is shuffled (MW)")
ax.set_title("Permutation Importance — Response Model (Model B)")
ax.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()

print(perm_importance.sort_values(ascending=False).round(1))
```

```python
# ============================================================
# PHASE 8 — CELL 5 — Temperature response curve (partial dependence)
# ============================================================

def response_curve(df, model, ref_columns, temps):
    """Set temperature to each value in `temps` for every row, holding all
    calendar/other-weather features at their real values, and average the
    predictions. This isolates the temperature-demand relationship the model
    has learned.
    """
    means = []
    for t in temps:
        X = build_response_features(df, ref_columns=ref_columns)
        X["temp"] = t
        X["cdd"] = max(t - BASE_TEMP, 0.0)
        X["hdd"] = max(BASE_TEMP - t, 0.0)
        means.append(model.predict(X).mean())
    return np.array(means)


# Sweep only within the observed training range — a Random Forest cannot
# extrapolate beyond the data it saw, so points outside are meaningless.
t_lo = np.floor(train_df["temp"].min())
t_hi = np.ceil(train_df["temp"].max())
temps_all = np.arange(t_lo, t_hi + 0.5, 0.5)

curve_all = response_curve(test_df, rf_response, train_Xr.columns, temps_all)

fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

# --- Panel 1: overall response curve ---
axes[0].plot(temps_all, curve_all, linewidth=2)
turning_point = temps_all[curve_all.argmin()]
axes[0].axvline(turning_point, color="red", linestyle="--",
                label=f"Minimum at {turning_point:.1f} °C")
axes[0].set_xlabel("Temperature (°C)")
axes[0].set_ylabel("Mean Predicted Demand (MW)")
axes[0].set_title("Temperature Response Curve (Model B, test period)")
axes[0].legend()
axes[0].grid(alpha=0.3)

# --- Panel 2: per season, each within its OWN observed range ---
# Sweeping winter rows up to 35 °C would be physically absurd, so we
# restrict each season's curve to the temperatures that season actually saw.
for season in ["winter", "spring", "summer", "autumn"]:
    sub = test_df[test_df["season"] == season]
    if len(sub) == 0:
        continue
    s_temps = np.arange(np.floor(sub["temp"].min()),
                        np.ceil(sub["temp"].max()) + 0.5, 0.5)
    s_curve = response_curve(sub, rf_response, train_Xr.columns, s_temps)
    axes[1].plot(s_temps, s_curve, linewidth=2, label=season)

axes[1].set_xlabel("Temperature (°C)")
axes[1].set_ylabel("Mean Predicted Demand (MW)")
axes[1].set_title("Temperature Response by Season (observed range only)")
axes[1].legend(title="Season")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Demand-minimising temperature: {turning_point:.1f} °C")
print(f"Demand at coldest ({temps_all[0]:.0f} °C): {curve_all[0]:,.0f} MW")
print(f"Demand at minimum:                {curve_all.min():,.0f} MW")
print(f"Demand at hottest ({temps_all[-1]:.0f} °C): {curve_all[-1]:,.0f} MW")
```

```python
# ============================================================
# PHASE 8 — CELL 6 — Counterfactual warming scenarios (+1/+2/+3 °C)
# ============================================================

scenario_preds = {
    "Baseline": rf_response.predict(
        build_response_features(test_df, ref_columns=train_Xr.columns)
    )
}

for delta in [1, 2, 3]:
    # temp_shift also recomputes cdd and hdd — see Cell 8.1.
    scenario_preds[f"+{delta}°C"] = rf_response.predict(
        build_response_features(test_df, temp_shift=delta,
                                ref_columns=train_Xr.columns)
    )

baseline_mean = scenario_preds["Baseline"].mean()

rows = []
for name, preds in scenario_preds.items():
    change_mw = preds.mean() - baseline_mean
    rows.append({
        "Scenario": name,
        "Mean Predicted Demand (MW)": preds.mean(),
        "Change (MW)": change_mw,
        "Change (%)": 100 * change_mw / baseline_mean,
    })

scenario_table = pd.DataFrame(rows)
print(scenario_table.round(3).to_string(index=False))

# Sanity check: only the three temperature-derived columns should differ.
X_base  = build_response_features(test_df, ref_columns=train_Xr.columns)
X_plus1 = build_response_features(test_df, temp_shift=1, ref_columns=train_Xr.columns)
differing = [c for c in X_base.columns if not X_base[c].equals(X_plus1[c])]
print("\nColumns that change between baseline and +1 °C:", differing)
print("(Should be exactly ['temp', 'cdd', 'hdd'])")
```

```python
# ============================================================
# PHASE 8 — CELL 7 — Where the effect actually shows up
# ============================================================

diff_df = pd.DataFrame({
    "season": test_df["season"].values,
    "hour": test_df["hour"].values,
    "baseline": scenario_preds["Baseline"],
})
for delta in [1, 2, 3]:
    diff_df[f"+{delta}°C"] = scenario_preds[f"+{delta}°C"] - scenario_preds["Baseline"]

scen_cols = ["+1°C", "+2°C", "+3°C"]
season_order = ["winter", "spring", "summer", "autumn"]

seasonal_mw = diff_df.groupby("season")[scen_cols].mean().reindex(season_order)
seasonal_baseline = diff_df.groupby("season")["baseline"].mean().reindex(season_order)
seasonal_pct = 100 * seasonal_mw.div(seasonal_baseline, axis=0)

print("Mean change in predicted demand (MW):")
print(seasonal_mw.round(1))
print("\nMean change in predicted demand (% of seasonal baseline):")
print(seasonal_pct.round(2))

fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

x = np.arange(len(season_order))
width = 0.25
for i, scen in enumerate(scen_cols):
    axes[0].bar(x + (i - 1) * width, seasonal_pct[scen], width, label=scen)
axes[0].axhline(0, color="black", linewidth=0.8)
axes[0].set_xticks(x)
axes[0].set_xticklabels(season_order)
axes[0].set_xlabel("Season")
axes[0].set_ylabel("Change in Predicted Demand (%)")
axes[0].set_title("Seasonal Demand Response to Warming")
axes[0].legend(title="Scenario")
axes[0].grid(axis="y", alpha=0.3)

# Hour-of-day profile under +3 °C — this is what matters for grid peak capacity.
for season in season_order:
    sub = diff_df[diff_df["season"] == season]
    hourly = sub.groupby("hour")["+3°C"].mean()
    axes[1].plot(hourly.index, hourly.values, marker="o", markersize=3, label=season)
axes[1].axhline(0, color="black", linewidth=0.8)
axes[1].set_xlabel("Hour of Day (local time)")
axes[1].set_ylabel("Change in Predicted Demand (MW)")
axes[1].set_title("Hourly Demand Response Under +3 °C")
axes[1].set_xticks(range(0, 24, 2))
axes[1].legend(title="Season")
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

summer_hourly = diff_df[diff_df["season"] == "summer"].groupby("hour")["+3°C"].mean()
print(f"\nLargest summer increase under +3 °C: {summer_hourly.max():,.0f} MW "
      f"at hour {summer_hourly.idxmax()}")
```

```python
# ============================================================
# PHASE 8 — CELL 8 — Are these changes distinguishable from zero?
# ============================================================

def block_bootstrap_ci(values, block_ids, n_boot=1000, seed=42):
    """Bootstrap the mean, resampling whole DAYS rather than individual hours.

    Hourly demand is strongly autocorrelated — consecutive hours are not
    independent observations. Resampling individual hours would pretend we
    have 8757 independent data points when we effectively have ~365, and
    would produce confidence intervals that are far too narrow.
    """
    rng = np.random.default_rng(seed)
    unique_blocks = pd.unique(block_ids)
    lookup = {b: values[block_ids == b] for b in unique_blocks}

    means = np.empty(n_boot)
    for i in range(n_boot):
        picked = rng.choice(unique_blocks, size=len(unique_blocks), replace=True)
        means[i] = np.concatenate([lookup[b] for b in picked]).mean()

    return means.mean(), np.percentile(means, 2.5), np.percentile(means, 97.5)


days = test_df["datetime_utc"].dt.date.values

ci_rows = []
for delta in [1, 2, 3]:
    m, lo, hi = block_bootstrap_ci(diff_df[f"+{delta}°C"].values, days)
    ci_rows.append({
        "Scenario": f"+{delta}°C",
        "Mean Change (MW)": m,
        "95% CI lower": lo,
        "95% CI upper": hi,
        "CI excludes zero": bool(lo > 0 or hi < 0),
    })

ci_all = pd.DataFrame(ci_rows)
print("ANNUAL MEAN — all hours:")
print(ci_all.round(2).to_string(index=False))

print("\nBY SEASON:")
season_ci_rows = []
for season in season_order:
    mask = (diff_df["season"] == season).values
    for delta in [1, 2, 3]:
        m, lo, hi = block_bootstrap_ci(diff_df.loc[mask, f"+{delta}°C"].values,
                                       days[mask])
        season_ci_rows.append({
            "Season": season,
            "Scenario": f"+{delta}°C",
            "Mean Change (MW)": m,
            "95% CI lower": lo,
            "95% CI upper": hi,
            "CI excludes zero": bool(lo > 0 or hi < 0),
        })

season_ci = pd.DataFrame(season_ci_rows)
print(season_ci.round(2).to_string(index=False))
```

```python
# ============================================================
# PHASE 8 — CELL 9 — How far outside the training range do we go?
# ============================================================

train_temp_max = train_df["temp"].max()
train_temp_min = train_df["temp"].min()

print(f"Training temperature range: {train_temp_min:.1f} °C to {train_temp_max:.1f} °C")
print(f"Test temperature range:     {test_df['temp'].min():.1f} °C to "
      f"{test_df['temp'].max():.1f} °C\n")

for delta in [1, 2, 3]:
    shifted = test_df["temp"] + delta
    above = (shifted > train_temp_max).mean()
    n_above = (shifted > train_temp_max).sum()
    print(f"+{delta} °C: {above:.2%} of hours ({n_above}) exceed the training maximum")

print("""
Interpretation:
A Random Forest predicts by averaging training observations. It CANNOT
extrapolate: for any temperature above the training maximum it returns the
same value it learned at the maximum. Every hour counted above is therefore
one where the warming effect is systematically UNDERSTATED.

Our estimates are best read as a LOWER BOUND on the warming response.
""")
```

```python
# ============================================================
# PHASE 8 — CELL 10 — Robustness: does the finding depend on our choices?
# ============================================================

# --- Check 1: month and season are collinear with temperature ------------
# July is hot AND is a holiday month with different industrial activity.
# Keeping 'month'/'season' lets the model attribute genuine temperature
# effects to the calendar instead, DAMPING the estimated response. Dropping
# them risks the opposite (attributing non-thermal seasonality to temp).
# Reporting both bounds the answer.
train_Xs = build_response_features(train_df, drop_calendar_season=True)
test_Xs  = build_response_features(test_df, ref_columns=train_Xs.columns,
                                   drop_calendar_season=True)

rf_nocal = RandomForestRegressor(n_estimators=300, min_samples_leaf=5,
                                 random_state=42, n_jobs=-1)
rf_nocal.fit(train_Xs, train_yr)

base_nocal = rf_nocal.predict(test_Xs)
print("Sensitivity 1 — model WITHOUT month/season features")
print(f"  R2: {r2_score(test_yr, base_nocal):.4f}")
for delta in [1, 2, 3]:
    p = rf_nocal.predict(
        build_response_features(test_df, temp_shift=delta,
                                ref_columns=train_Xs.columns,
                                drop_calendar_season=True)
    )
    change = p.mean() - base_nocal.mean()
    print(f"  +{delta} °C: {change:+8.1f} MW "
          f"({100 * change / base_nocal.mean():+.3f}%)")

# --- Check 2: is 18 °C the right balance point? --------------------------
print("\nSensitivity 2 — balance-point temperature")
original_base = BASE_TEMP
for candidate in [16.0, 18.0, 20.0]:
    BASE_TEMP = candidate
    Xtr = build_response_features(train_df)
    Xte = build_response_features(test_df, ref_columns=Xtr.columns)
    m = RandomForestRegressor(n_estimators=200, min_samples_leaf=5,
                              random_state=42, n_jobs=-1).fit(Xtr, train_yr)
    b = m.predict(Xte)
    p3 = m.predict(build_response_features(test_df, temp_shift=3,
                                           ref_columns=Xtr.columns))
    print(f"  base {candidate:.0f} °C -> R2 {r2_score(test_yr, b):.4f}, "
          f"+3 °C change {p3.mean() - b.mean():+.1f} MW")
BASE_TEMP = original_base   # restore before continuing
```

```python
# ============================================================
# PHASE 10 — CELL 1 — Build hourly fossil CO2 from generation data
# ============================================================

# ####################################################################
# EXTERNAL ASSUMPTION — YOU MUST REPLACE THESE WITH CITED VALUES.
#
# Direct-combustion CO2 per kWh of ELECTRICITY GENERATED (kg CO2/kWh).
# The values below are commonly cited central estimates, NOT figures I
# can source for you. Before submitting, look each one up and cite it.
# Suitable sources: IPCC AR5 Working Group III Annex III (Table A.III.2),
# the European Environment Agency's electricity emission factor reports,
# or Spain's National Inventory Report to the UNFCCC.
#
# Record the LOW and HIGH of the published range too — Cell 10.4 uses
# them for a sensitivity band.
# ####################################################################

EMISSION_FACTORS = {
    "generation fossil gas":                 0.40,   # CCGT — VERIFY
    "generation fossil hard coal":           0.90,   # VERIFY
    "generation fossil brown coal/lignite":  1.05,   # VERIFY
    "generation fossil oil":                 0.70,   # VERIFY
    "generation fossil oil shale":           1.10,   # VERIFY
    "generation fossil peat":                1.05,   # VERIFY
    "generation fossil coal-derived gas":    0.90,   # VERIFY
}

# Nuclear, hydro, wind, solar and geothermal are treated as zero DIRECT
# combustion emissions. Lifecycle emissions are non-zero but out of scope —
# state this in Limitations.
#
# Biomass and waste are a judgement call: biogenic CO2 is conventionally
# reported as zero under IPCC accounting, but this is contested. We treat
# them as zero and flag it. If you disagree, add them with a citation.

present = {c: f for c, f in EMISSION_FACTORS.items() if c in merged.columns}
absent  = [c for c in EMISSION_FACTORS if c not in merged.columns]

print("Fossil columns found in your data:")
for c in present:
    print(f"  {c}")
if absent:
    print("\nNot present (dropped in Phase 2 as entirely empty):")
    for c in absent:
        print(f"  {c}")

gen = merged[list(present)].fillna(0.0)

always_zero = [c for c in present if gen[c].abs().sum() == 0]
if always_zero:
    print("\nPresent but always zero (no contribution to emissions):")
    for c in always_zero:
        print(f"  {c}")

# MW of generation -> kg CO2 per hour:
#   MW x 1 h = MWh;  MWh x 1000 = kWh;  kWh x factor = kg CO2
merged["fossil_co2_kg_h"] = sum(gen[c] * 1000 * f for c, f in present.items())

total_co2_t = merged["fossil_co2_kg_h"].sum() / 1000
total_load_mwh = merged["total load actual"].sum()
implied_average = merged["fossil_co2_kg_h"].sum() / (total_load_mwh * 1000)

print(f"\nTotal fossil CO2 over the dataset: {total_co2_t:,.0f} tonnes")
print(f"Implied AVERAGE emission factor:   {implied_average:.3f} kg CO2/kWh")
print(f"(Compare with the 0.321 kg/kWh national figure you had before.)")

print("\nShare of hours with zero fossil generation: "
      f"{(merged['fossil_co2_kg_h'] == 0).mean():.2%}")
```

```python
# ============================================================
# PHASE 10 — CELL 2 — Marginal emission factor by regression
# ============================================================

# Method: regress the HOUR-TO-HOUR CHANGE in fossil CO2 against the
# hour-to-hour CHANGE in total demand. The slope answers "when demand
# rises by 1 MWh, how much extra CO2 is emitted?" — which is precisely
# the marginal factor. Working in differences also removes slow-moving
# confounders (fuel prices, capacity additions, seasonal patterns)
# because they barely change from one hour to the next.
#
# This approach follows the published marginal-emission-factor literature
# (e.g. Hawkes 2010; Siler-Evans, Azevedo & Morgan 2012) — cite one of
# these, and note it as a method taken from outside the course syllabus.

work = merged[["datetime_utc", "total load actual", "fossil_co2_kg_h"]].copy()
work["d_load"] = work["total load actual"].diff()          # MW (= MWh over 1 h)
work["d_co2"]  = work["fossil_co2_kg_h"].diff()            # kg CO2

# Only use genuinely consecutive hours — a gap would make the difference
# meaningless.
work["gap_h"] = work["datetime_utc"].diff().dt.total_seconds() / 3600
work = work[(work["gap_h"] == 1) & work["d_load"].notna() & work["d_co2"].notna()]

# Fit through the origin: zero change in demand should mean zero change
# in emissions.
slope_origin = (work["d_load"] * work["d_co2"]).sum() / (work["d_load"] ** 2).sum()

# Also fit with an intercept, as a diagnostic. A large intercept would
# mean emissions drift independently of demand.
slope_int, intercept = np.polyfit(work["d_load"], work["d_co2"], 1)
r2_marginal = r2_score(work["d_co2"], slope_int * work["d_load"] + intercept)

MARGINAL_FACTOR = slope_origin / 1000    # kg CO2 per kWh

print(f"Consecutive-hour pairs used: {len(work):,}")
print(f"\nMarginal factor (through origin): {MARGINAL_FACTOR:.3f} kg CO2/kWh")
print(f"Marginal factor (with intercept): {slope_int / 1000:.3f} kg CO2/kWh")
print(f"  intercept: {intercept:,.0f} kg/h  (should be small vs typical |d_co2|)")
print(f"  R2 of the difference regression: {r2_marginal:.3f}")
print(f"\nFor comparison — average factor: {implied_average:.3f} kg CO2/kWh")
print(f"National average you used before:  0.321 kg CO2/kWh")

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(work["d_load"], work["d_co2"] / 1000, alpha=0.05, s=4)
xs = np.linspace(work["d_load"].min(), work["d_load"].max(), 100)
ax.plot(xs, slope_origin * xs / 1000, color="red", linewidth=1.5,
        label=f"Marginal fit: {MARGINAL_FACTOR:.3f} kg/kWh")
ax.axhline(0, color="black", linewidth=0.5)
ax.axvline(0, color="black", linewidth=0.5)
ax.set_xlabel("Hour-to-hour change in demand (MW)")
ax.set_ylabel("Hour-to-hour change in fossil CO₂ (tonnes/hour)")
ax.set_title("Marginal Emission Factor Estimated from Hourly Data")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PHASE 10 — CELL 5 — Which generators actually follow demand?
# ============================================================

# Same differencing logic as Cell 10.2, but applied to every generation
# source separately. The slope for each source is "MW of this source per
# MW of demand change" — i.e. its share of the marginal response.

gen_cols = [c for c in merged.columns if c.startswith("generation ")]

w = merged[["datetime_utc", "total load actual"] + gen_cols].copy()
gap_h = w["datetime_utc"].diff().dt.total_seconds() / 3600

d = w[gen_cols + ["total load actual"]].diff()
d["gap_h"] = gap_h
d = d[(d["gap_h"] == 1) & d["total load actual"].notna()]

denom = (d["total load actual"] ** 2).sum()

rows = []
for c in gen_cols:
    if merged[c].fillna(0).abs().sum() == 0:
        continue                      # always-zero column, skip
    slope = (d["total load actual"] * d[c].fillna(0.0)).sum() / denom
    rows.append({
        "Source": c.replace("generation ", ""),
        "MW per MW of demand change": slope,
        "Mean output (MW)": merged[c].mean(),
    })

dispatch = (pd.DataFrame(rows)
            .sort_values("MW per MW of demand change", ascending=False))
total_slope = dispatch["MW per MW of demand change"].sum()
dispatch["Share of response (%)"] = (
    100 * dispatch["MW per MW of demand change"] / total_slope
)

print(dispatch.round(3).to_string(index=False))
print(f"\nSlopes sum to: {total_slope:.3f}")

# Cross-check: rebuild the marginal factor from the decomposition and
# compare with Cell 10.2's regression estimate. They should agree closely —
# they are two routes to the same quantity.
implied_marginal = sum(
    dispatch.loc[dispatch["Source"] == c.replace("generation ", ""),
                 "MW per MW of demand change"].sum() * f
    for c, f in EMISSION_FACTORS.items()
)
fossil_share = sum(
    dispatch.loc[dispatch["Source"] == c.replace("generation ", ""),
                 "MW per MW of demand change"].sum()
    for c in EMISSION_FACTORS
)

print(f"\nFossil share of the marginal response: {fossil_share:.3f}")
print(f"Zero-carbon share:                     {total_slope - fossil_share:.3f}")
print(f"\nMarginal factor from decomposition: {implied_marginal:.3f} kg/kWh")
print(f"Marginal factor from Cell 10.2:     {MARGINAL_FACTOR:.3f} kg/kWh")
print("(These should be close — if not, tell me.)")

fig, ax = plt.subplots(figsize=(9, 7))
top = dispatch[dispatch["MW per MW of demand change"].abs() > 0.005]
colors = ["#D85A30" if s.startswith("fossil") else "#1D9E75"
          for s in top["Source"]]
ax.barh(top["Source"], top["MW per MW of demand change"], color=colors)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_xlabel("MW of generation per MW of demand change")
ax.set_title("Which Sources Follow Demand? (fossil in orange, zero-carbon in green)")
ax.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PHASE 10 — CELL 6 — Piecewise degree bands (convexity test)
# ============================================================

# Instead of one "degrees above 18" variable, split the temperature range
# into bands and give each its own coefficient. Each hour contributes its
# degrees to whichever bands it spans:
#   a 24 °C hour contributes 4 degrees to the 18-22 band and 2 to 22-26.
# The coefficient on each band is then MW per °C WITHIN that band.

CDD_EDGES = [18, 22, 26, 30]        # cooling bands, last is open-ended
HDD_EDGES = [18, 14, 10, 6]         # heating bands, last is open-ended


def degree_bands(t):
    out = {}
    for i, lo in enumerate(CDD_EDGES):
        hi = CDD_EDGES[i + 1] if i + 1 < len(CDD_EDGES) else None
        v = (t - lo).clip(lower=0)
        name = f"cdd_{lo}_{hi}" if hi else f"cdd_{lo}_plus"
        out[name] = v.clip(upper=hi - lo) if hi else v
    for i, hi in enumerate(HDD_EDGES):
        lo = HDD_EDGES[i + 1] if i + 1 < len(HDD_EDGES) else None
        v = (hi - t).clip(lower=0)
        name = f"hdd_{hi}_{lo}" if lo is not None else f"hdd_{hi}_minus"
        out[name] = v.clip(upper=hi - lo) if lo is not None else v
    return out


def build_piecewise(df, temp_shift=0.0, ref_columns=None):
    s = df["temp"] + temp_shift
    X = pd.DataFrame(degree_bands(s), index=df.index)
    X["humidity"] = df["humidity"]
    X["wind_speed"] = df["wind_speed"]
    for col, prefix in [("hour", "h"), ("day_of_week", "dow"), ("month", "m")]:
        dm = pd.get_dummies(df[col].astype(int), prefix=prefix,
                            drop_first=True).astype(int)
        X = pd.concat([X, dm], axis=1)
    if ref_columns is not None:
        X = X.reindex(columns=ref_columns, fill_value=0)
    return X


train_Xp = build_piecewise(train_df)
test_Xp = build_piecewise(test_df, ref_columns=train_Xp.columns)

pw_model = LinearRegression().fit(train_Xp, train_yr)
pw_pred = pw_model.predict(test_Xp)

print("Model comparison (all on the same test set):")
print(f"  Random Forest:     R2 {r2_score(test_yr, pred_response):.4f}")
print(f"  Linear (1 slope):  R2 {r2_score(test_yr, pred_lin):.4f}")
print(f"  Linear (piecewise): R2 {r2_score(test_yr, pw_pred):.4f}, "
      f"MAE {mean_absolute_error(test_yr, pw_pred):.1f} MW")

band_cols = [c for c in train_Xp.columns if c.startswith(("cdd_", "hdd_"))]
band_coefs = pd.Series(pw_model.coef_, index=train_Xp.columns)[band_cols]

print("\n--- MW PER DEGREE C WITHIN EACH BAND ---")
print(band_coefs.round(1))

cdd_bands = [c for c in band_cols if c.startswith("cdd_")]
print(f"\nCooling slope, coolest band: {band_coefs[cdd_bands[0]]:.1f} MW/°C")
print(f"Cooling slope, hottest band: {band_coefs[cdd_bands[-1]]:.1f} MW/°C")
if band_coefs[cdd_bands[0]] != 0:
    print(f"Steepening factor: {band_coefs[cdd_bands[-1]] / band_coefs[cdd_bands[0]]:.1f}x")

# Re-run the scenarios with this model — a third independent estimate.
pw_base = pw_model.predict(test_Xp)
pw_rows = []
for delta in [1, 2, 3]:
    p = pw_model.predict(build_piecewise(test_df, temp_shift=delta,
                                         ref_columns=train_Xp.columns))
    s = (pd.DataFrame({"season": test_df["season"].values, "d": p - pw_base})
         .groupby("season")["d"].mean())
    pw_rows.append(s.rename(f"+{delta}°C"))
pw_seasonal = pd.concat(pw_rows, axis=1).reindex(season_order)

print("\n--- THREE-MODEL COMPARISON, seasonal change (MW) ---")
print("\nRandom Forest:")
print(rf_seasonal.round(1))
print("\nLinear, single slope:")
print(lin_seasonal.round(1))
print("\nLinear, piecewise bands:")
print(pw_seasonal.round(1))

fig, ax = plt.subplots(figsize=(9, 5.5))
x_pos = np.arange(len(cdd_bands))
ax.bar(x_pos, band_coefs[cdd_bands].values, color="#D85A30", label="Cooling")
ax.set_xticks(x_pos)
ax.set_xticklabels([c.replace("cdd_", "").replace("_", "–") + " °C"
                    for c in cdd_bands])
ax.set_xlabel("Temperature band")
ax.set_ylabel("Additional demand per °C within band (MW)")
ax.set_title("Does the Cooling Response Steepen with Temperature?")
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PHASE 10 — CELL 7 — Does the cooling response depend on time of day?
# ============================================================

# Four time blocks rather than 24 hourly interactions: with 24 we'd have
# 48 interaction terms on sparse cells and would reproduce exactly the
# collinearity that wrecked Cell 6's band coefficients. Four blocks keeps
# each cell well populated.
TIME_BLOCKS = {
    "night":     range(0, 8),
    "morning":   range(8, 14),
    "afternoon": range(14, 20),
    "evening":   range(20, 24),
}


def block_of(h):
    for name, hours in TIME_BLOCKS.items():
        if h in hours:
            return name
    return "night"


def build_interaction(df, temp_shift=0.0, ref_columns=None):
    """Linear features plus CDD/HDD interacted with time-of-day block.

    'cdd' keeps its main effect and each block gets a DEVIATION from it,
    so the cooling slope in block b is coef['cdd'] + coef['cdd_x_b'].
    """
    shifted = df["temp"] + temp_shift
    cdd = (shifted - BASE_TEMP).clip(lower=0)
    hdd = (BASE_TEMP - shifted).clip(lower=0)
    blocks = df["hour"].map(block_of)

    X = pd.DataFrame(index=df.index)
    X["cdd"] = cdd
    X["hdd"] = hdd
    for b in TIME_BLOCKS:
        ind = (blocks == b).astype(int)
        X[f"cdd_x_{b}"] = cdd * ind
        X[f"hdd_x_{b}"] = hdd * ind

    X["humidity"] = df["humidity"]
    X["wind_speed"] = df["wind_speed"]
    for col, prefix in [("hour", "h"), ("day_of_week", "dow"), ("month", "m")]:
        dm = pd.get_dummies(df[col].astype(int), prefix=prefix,
                            drop_first=True).astype(int)
        X = pd.concat([X, dm], axis=1)

    if ref_columns is not None:
        X = X.reindex(columns=ref_columns, fill_value=0)
    return X


train_Xi = build_interaction(train_df)
test_Xi = build_interaction(test_df, ref_columns=train_Xi.columns)

int_model = LinearRegression().fit(train_Xi, train_yr)
int_pred = int_model.predict(test_Xi)
r2_int = r2_score(test_yr, int_pred)

print("MODEL COMPARISON (same test set)")
print(f"  Linear, single slope:   R2 {r2_score(test_yr, pred_lin):.4f}")
print(f"  Linear, piecewise:      R2 {r2_score(test_yr, pw_pred):.4f}")
print(f"  Linear + interactions:  R2 {r2_int:.4f}   "
      f"MAE {mean_absolute_error(test_yr, int_pred):.1f} MW")
print(f"  Random Forest:          R2 {r2_score(test_yr, pred_response):.4f}")
print(f"\nR2 gain from interactions: {r2_int - r2_score(test_yr, pred_lin):+.4f}")

coefs_int = pd.Series(int_model.coef_, index=train_Xi.columns)

print("\n--- COOLING SLOPE BY TIME OF DAY (MW per °C above 18 °C) ---")
for b in TIME_BLOCKS:
    print(f"  {b:10s} {coefs_int['cdd'] + coefs_int[f'cdd_x_{b}']:8.1f}")

print("\n--- HEATING SLOPE BY TIME OF DAY (MW per °C below 18 °C) ---")
for b in TIME_BLOCKS:
    print(f"  {b:10s} {coefs_int['hdd'] + coefs_int[f'hdd_x_{b}']:8.1f}")

# Re-run the scenarios with this model.
int_base = int_model.predict(test_Xi)
int_rows = []
for delta in [1, 2, 3]:
    p = int_model.predict(build_interaction(test_df, temp_shift=delta,
                                            ref_columns=train_Xi.columns))
    s = (pd.DataFrame({"season": test_df["season"].values, "d": p - int_base})
         .groupby("season")["d"].mean())
    int_rows.append(s.rename(f"+{delta}°C"))
int_seasonal = pd.concat(int_rows, axis=1).reindex(season_order)

print("\n--- FOUR-MODEL COMPARISON, seasonal change (MW) ---")
comparison = pd.DataFrame({
    "Linear (1 slope)":    lin_seasonal["+3°C"],
    "Linear (piecewise)":  pw_seasonal["+3°C"],
    "Linear + interact":   int_seasonal["+3°C"],
    "Random Forest":       rf_seasonal["+3°C"],
})
print("\n+3 °C effect by season:")
print(comparison.round(1))

print("\nFull interaction-model results:")
print(int_seasonal.round(1))
print(f"\nAnnual mean change: "
      f"{(int_model.predict(build_interaction(test_df, temp_shift=3, ref_columns=train_Xi.columns)) - int_base).mean():+.1f} MW")

fig, ax = plt.subplots(figsize=(10, 5.5))
blocks = list(TIME_BLOCKS)
cool = [coefs_int["cdd"] + coefs_int[f"cdd_x_{b}"] for b in blocks]
heat = [coefs_int["hdd"] + coefs_int[f"hdd_x_{b}"] for b in blocks]
xb = np.arange(len(blocks))
ax.bar(xb - 0.2, cool, 0.4, label="Cooling (above 18 °C)", color="#D85A30")
ax.bar(xb + 0.2, heat, 0.4, label="Heating (below 18 °C)", color="#378ADD")
ax.axhline(0, color="black", linewidth=0.8)
ax.set_xticks(xb)
ax.set_xticklabels([f"{b}\n{list(TIME_BLOCKS[b])[0]}–{list(TIME_BLOCKS[b])[-1]}h"
                    for b in blocks])
ax.set_xlabel("Time of day (local)")
ax.set_ylabel("Demand sensitivity (MW per °C)")
ax.set_title("Temperature Sensitivity of Demand by Time of Day")
ax.legend()
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

#Phase 11 was a later addittion after seeing the results in phase 9 and phase 10.

###Cell 8.8's bootstrap resampled which hours you average over. That is a real source of uncertainty, but it is not the main one and for a linear model it is almost nothing. The per-hour effect of +1 °C in a linear model is a fixed function of the coefficients, so within winter (where nearly every hour sits below the balance point) the effect is close to constant from hour to hour. Resampling hours therefore produces an interval of ±0.1%, which is not precision, it's a blind spot.

###What it never captured is uncertainty in the fitted coefficients themselves. Your cooling slope of ~175 MW/°C is an estimate from finite data; a different four years would have given a slightly different number, and the scenario effect inherits that.

###The forest partly concealed this, because its per-hour effects are jumpy step functions and hour-resampling produced plausible-looking width by coincidence. Neither model's original intervals were measuring the right quantity.

###The fix below resamples training days, refits the model, and recomputes the effect several hundred times. The spread across refits is the coefficient uncertainty.

```python
# ============================================================
# PHASE 11 — CELL 1 — Primary model = linear + interactions
# ============================================================

# Preserve the Random Forest results before overwriting anything —
# they become the robustness check in the paper.
rf_diff_df   = diff_df.copy()
rf_ci_all    = ci_all.copy()
rf_season_ci = season_ci.copy()
rf_seasonal_saved = rf_seasonal.copy()

# Chosen on four grounds:
#   1. Best test R2 of all four specifications (0.6351) and lowest MAE.
#   2. Interpretable — coefficients are MW per degree C by time of day.
#   3. Extrapolates above the training temperature range; the forest cannot.
#   4. Encodes a mechanism we can defend: heating sensitivity depends on
#      occupancy, so it varies by time of day.
PRIMARY_MODEL   = int_model
PRIMARY_BUILDER = build_interaction
PRIMARY_REF     = train_Xi.columns
PRIMARY_NAME    = "Linear + time-of-day interactions"

print(f"Primary model: {PRIMARY_NAME}")
print(f"  Test R2:  {r2_score(test_yr, int_pred):.4f}")
print(f"  Test MAE: {mean_absolute_error(test_yr, int_pred):.1f} MW")
print(f"  Features: {len(PRIMARY_REF)}")

# Regenerate the scenario predictions and per-hour differences.
scenario_preds = {"Baseline": PRIMARY_MODEL.predict(
    PRIMARY_BUILDER(test_df, ref_columns=PRIMARY_REF))}
for delta in [1, 2, 3]:
    scenario_preds[f"+{delta}°C"] = PRIMARY_MODEL.predict(
        PRIMARY_BUILDER(test_df, temp_shift=delta, ref_columns=PRIMARY_REF))

diff_df = pd.DataFrame({
    "season":   test_df["season"].values,
    "hour":     test_df["hour"].values,
    "baseline": scenario_preds["Baseline"],
})
for delta in [1, 2, 3]:
    diff_df[f"+{delta}°C"] = scenario_preds[f"+{delta}°C"] - scenario_preds["Baseline"]

print("\nSeasonal effects (point estimates, MW):")
print(diff_df.groupby("season")[scen_cols].mean().reindex(season_order).round(1))
```

```python
# ============================================================
# PHASE 11 — CELL 2 — Bootstrap that captures COEFFICIENT uncertainty
# ============================================================
# Runtime: roughly 20-60 seconds in Colab for 300 refits.

import time

# Convert to numpy once — refitting 300 times on DataFrames is needlessly slow,
# and rebuilding the feature matrices each iteration would be slower still.
# The test matrices never change, so build them once.
Xtr = PRIMARY_BUILDER(train_df, ref_columns=PRIMARY_REF).to_numpy(dtype=float)
ytr = train_yr.to_numpy(dtype=float)

Xte = {0: PRIMARY_BUILDER(test_df, ref_columns=PRIMARY_REF).to_numpy(dtype=float)}
for delta in [1, 2, 3]:
    Xte[delta] = PRIMARY_BUILDER(test_df, temp_shift=delta,
                                 ref_columns=PRIMARY_REF).to_numpy(dtype=float)

# Resample whole TRAINING days, not individual hours — consecutive hours are
# not independent observations, so hour-level resampling would understate
# uncertainty for the same reason it did before.
train_days = train_df["datetime_utc"].dt.date.to_numpy()
unique_days = pd.unique(train_days)
day_rows = {d: np.flatnonzero(train_days == d) for d in unique_days}

test_season = test_df["season"].to_numpy()
season_masks = {s: (test_season == s) for s in season_order}
scopes = season_order + ["annual"]


def effects_from_model(model):
    base = model.predict(Xte[0])
    out = {}
    for delta in [1, 2, 3]:
        diff = model.predict(Xte[delta]) - base
        out[delta] = {s: diff[m].mean() for s, m in season_masks.items()}
        out[delta]["annual"] = diff.mean()
    return out


point_effects = effects_from_model(PRIMARY_MODEL)

N_BOOT = 300
boot = {d: {s: np.empty(N_BOOT) for s in scopes} for d in [1, 2, 3]}

rng_boot = np.random.default_rng(42)
t0 = time.time()
for b in range(N_BOOT):
    picked = rng_boot.choice(unique_days, size=len(unique_days), replace=True)
    rows = np.concatenate([day_rows[d] for d in picked])
    m = LinearRegression().fit(Xtr[rows], ytr[rows])
    e = effects_from_model(m)
    for delta in [1, 2, 3]:
        for s in scopes:
            boot[delta][s][b] = e[delta][s]

print(f"{N_BOOT} refits in {time.time() - t0:.1f}s\n")

records = []
for delta in [1, 2, 3]:
    for s in scopes:
        arr = boot[delta][s]
        lo, hi = np.percentile(arr, [2.5, 97.5])
        records.append({
            "Season": s, "Scenario": f"+{delta}°C",
            "Mean Change (MW)": point_effects[delta][s],
            "95% CI lower": lo, "95% CI upper": hi,
            "CI excludes zero": bool(lo > 0 or hi < 0),
        })

boot_all = pd.DataFrame(records)

# Split into the two objects Phases 9 and 10 consume, keeping the same
# column names so none of that code needs editing.
ci_all = (boot_all[boot_all["Season"] == "annual"]
          .drop(columns=["Season"]).reset_index(drop=True))
season_ci = (boot_all[boot_all["Season"] != "annual"].reset_index(drop=True))

print("ANNUAL MEAN")
print(ci_all.round(2).to_string(index=False))
print("\nBY SEASON")
print(season_ci.round(2).to_string(index=False))

print("\n--- INTERVAL WIDTH: old method vs new ---")
for delta in [1, 2, 3]:
    scen = f"+{delta}°C"
    old = rf_season_ci[(rf_season_ci["Season"] == "summer") &
                       (rf_season_ci["Scenario"] == scen)].iloc[0]
    new = season_ci[(season_ci["Season"] == "summer") &
                    (season_ci["Scenario"] == scen)].iloc[0]
    print(f"  summer {scen}: "
          f"RF/hour-resample width {old['95% CI upper'] - old['95% CI lower']:7.1f} MW | "
          f"refit width {new['95% CI upper'] - new['95% CI lower']:7.1f} MW")
```

```python

# ============================================================
# PHASE 11 — CELL 3 — Learning curves (rubric requirement)
# ============================================================

from sklearn.model_selection import TimeSeriesSplit, learning_curve

# TimeSeriesSplit, not KFold: ordinary cross-validation would train on
# future data to predict the past. TimeSeriesSplit always trains on an
# earlier block and validates on the block immediately after it.
tscv = TimeSeriesSplit(n_splits=4)
train_sizes = np.linspace(0.2, 1.0, 5)

# Rebuild the primary design matrix on the TRAINING period only — the
# learning curve is an internal-validation diagnostic and must not touch
# the 2018 test set.
X_lc = PRIMARY_BUILDER(train_df, ref_columns=PRIMARY_REF)
y_lc = train_yr.to_numpy(dtype=float)

# Forest reduced to 100 trees here purely for runtime: the curve refits
# it 20 times (5 sizes x 4 folds). Shape is what matters, not the last
# decimal place.
curve_models = {
    "Linear + interactions (primary)": LinearRegression(),
    "Random forest": RandomForestRegressor(
        n_estimators=100, min_samples_leaf=5, random_state=42, n_jobs=-1),
}

curves = {}
for name, mdl in curve_models.items():
    sizes, train_scores, val_scores = learning_curve(
        mdl, X_lc, y_lc,
        train_sizes=train_sizes,
        cv=tscv,
        scoring="neg_mean_absolute_error",   # negated: sklearn maximises
        shuffle=False,                       # never shuffle a time series
        n_jobs=1,
    )
    curves[name] = (sizes, -train_scores, -val_scores)

    tr_final = (-train_scores).mean(axis=1)[-1]
    va_final = (-val_scores).mean(axis=1)[-1]
    print(f"{name}")
    print(f"  final training MAE:   {tr_final:8.1f} MW")
    print(f"  final validation MAE: {va_final:8.1f} MW")
    print(f"  generalisation gap:   {va_final - tr_final:8.1f} MW\n")

fig, axes = plt.subplots(1, len(curves), figsize=(7 * len(curves), 5), squeeze=False)
for ax, (name, (sizes, tr, va)) in zip(axes[0], curves.items()):
    tr_m, tr_s = tr.mean(axis=1), tr.std(axis=1)
    va_m, va_s = va.mean(axis=1), va.std(axis=1)

    ax.plot(sizes, tr_m, marker="o", label="Training MAE")
    ax.plot(sizes, va_m, marker="s", label="Validation MAE")
    ax.fill_between(sizes, tr_m - tr_s, tr_m + tr_s, alpha=0.15)
    ax.fill_between(sizes, va_m - va_s, va_m + va_s, alpha=0.15)

    ax.set_xlabel("Training set size (hours)")
    ax.set_ylabel("Mean Absolute Error (MW)")
    ax.set_title(name)
    ax.legend()
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PHASE 9 — CELL 1 — Classify each season/scenario using its CI
# ============================================================

# The decision rule is statistical, not an arbitrary MW cutoff:
#   CI contains zero          -> No detectable change
#   CI entirely above zero    -> Higher demand
#   CI entirely below zero    -> Lower demand
# This is why Cell 8.8 existed. Your original 0.01 MW threshold made
# "no change" impossible and labelled effects far below the model's own
# error as real findings.

season_baseline = diff_df.groupby("season")["baseline"].mean()


def classify_row(row):
    if not row["CI excludes zero"]:
        return "No detectable change"
    return "Higher demand" if row["Mean Change (MW)"] > 0 else "Lower demand"


classified = season_ci.copy()
classified["Change (%)"] = (
    100 * classified["Mean Change (MW)"] / classified["Season"].map(season_baseline)
)
classified["Classification"] = classified.apply(classify_row, axis=1)

cols = ["Season", "Scenario", "Mean Change (MW)", "95% CI lower", "95% CI upper",
        "Change (%)", "Classification"]
print(classified[cols].round(2).to_string(index=False))

print("\n--- CLASSIFICATION GRID ---")
grid = classified.pivot(index="Season", columns="Scenario",
                        values="Classification").reindex(season_order)
print(grid[scen_cols])

print("\n--- CLASS COUNTS ---")
print(classified["Classification"].value_counts())
```

```python
# ============================================================
# PHASE 9 — CELL 2 — Seasonal effects with 95% confidence intervals
# ============================================================

class_colors = {
    "Higher demand": "#D85A30",
    "Lower demand": "#378ADD",
    "No detectable change": "#888780",
}

fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))

x = np.arange(len(season_order))
width = 0.25

for i, scen in enumerate(scen_cols):
    sub = (classified[classified["Scenario"] == scen]
           .set_index("Season").reindex(season_order))

    # Error bars must be DISTANCES from the bar top, not absolute positions.
    lower_err = sub["Mean Change (MW)"] - sub["95% CI lower"]
    upper_err = sub["95% CI upper"] - sub["Mean Change (MW)"]

    axes[0].bar(
        x + (i - 1) * width, sub["Mean Change (MW)"], width,
        color=[class_colors[c] for c in sub["Classification"]],
        edgecolor="black", linewidth=0.4,
        yerr=[lower_err.values, upper_err.values], capsize=3,
        error_kw={"linewidth": 0.9},
        label=scen if i == 0 else None,
    )
    # Label each group with its warming level.
    for xi in x:
        axes[0].text(xi + (i - 1) * width, 0, scen, ha="center", va="bottom"
                     if sub["Mean Change (MW)"].iloc[xi] < 0 else "top",
                     fontsize=7, rotation=90, alpha=0.6)

axes[0].axhline(0, color="black", linewidth=0.8)
axes[0].set_xticks(x)
axes[0].set_xticklabels(season_order)
axes[0].set_xlabel("Season")
axes[0].set_ylabel("Change in Predicted Demand (MW)")
axes[0].set_title("Seasonal Demand Response with 95% Confidence Intervals")
axes[0].grid(axis="y", alpha=0.3)

handles = [plt.Rectangle((0, 0), 1, 1, color=c, ec="black", lw=0.4)
           for c in class_colors.values()]
axes[0].legend(handles, class_colors.keys(), fontsize=8, loc="best")

# --- Right panel: percentage change, same information, normalised ---
pct = classified.pivot(index="Season", columns="Scenario",
                       values="Change (%)").reindex(season_order)[scen_cols]
for i, scen in enumerate(scen_cols):
    axes[1].bar(x + (i - 1) * width, pct[scen], width, label=scen)
axes[1].axhline(0, color="black", linewidth=0.8)
axes[1].set_xticks(x)
axes[1].set_xticklabels(season_order)
axes[1].set_xlabel("Season")
axes[1].set_ylabel("Change in Predicted Demand (%)")
axes[1].set_title("Seasonal Demand Response (% of seasonal baseline)")
axes[1].legend(title="Scenario")
axes[1].grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# PHASE 9 — CELL 3 — The redistribution finding
# ============================================================

print("=" * 66)
print("ANNUAL TOTAL vs SEASONAL REDISTRIBUTION")
print("=" * 66)

for delta in [1, 2, 3]:
    scen = f"+{delta}°C"
    ann = ci_all[ci_all["Scenario"] == scen].iloc[0]
    sub = classified[classified["Scenario"] == scen]

    increases = sub.loc[sub["Mean Change (MW)"] > 0, "Mean Change (MW)"].sum()
    decreases = sub.loc[sub["Mean Change (MW)"] < 0, "Mean Change (MW)"].sum()

    print(f"\n{scen}")
    print(f"  Annual mean change:  {ann['Mean Change (MW)']:+8.1f} MW  "
          f"[{ann['95% CI lower']:+.1f}, {ann['95% CI upper']:+.1f}]  "
          f"detectable: {ann['CI excludes zero']}")
    print(f"  Seasonal increases:  {increases:+8.1f} MW")
    print(f"  Seasonal decreases:  {decreases:+8.1f} MW")
    print(f"  Gross movement:      {increases - decreases:8.1f} MW")

print("""
INTERPRETATION

The annual mean is not statistically distinguishable from zero, while the
seasonal effects are. These are not contradictory results — they are the
finding. Warming does not change how much electricity Spain uses over a
year so much as WHEN it uses it: summer demand rises, winter demand falls,
and the two largely cancel in the annual mean.

This matters for planning because generation and network capacity are sized
by PEAK demand, not annual energy. A redistribution toward summer afternoons
and evenings increases required capacity even when annual consumption is
unchanged.
""")
```

# PHASE 10 (continued) — CO₂ IMPACT OF THE SCENARIOS

These two cells depend on `classified` from Phase 9, so they run after it.
The emission factors themselves were established in Phase 10 Cells 1, 2 and 5 above.


```python
# ============================================================
# PHASE 10 — CELL 3 — Seasonal CO2 impact with uncertainty
# ============================================================

# Hours per season in the test period — needed to turn a mean MW change
# into a seasonal energy total.
season_hours = test_df["season"].value_counts().reindex(season_order)
print("Test hours per season:")
print(season_hours, "\n")


def mw_to_tonnes(mw_change, hours, factor):
    """Mean MW change over `hours` hours -> tonnes CO2.

    mw_change x hours = MWh;  MWh x 1000 = kWh;  x kg/kWh = kg;  /1000 = t.
    The two 1000s cancel, so MWh x factor = tonnes directly.
    """
    return mw_change * hours * factor


co2_rows = []
for _, row in classified.iterrows():
    hours = season_hours[row["Season"]]
    co2_rows.append({
        "Season": row["Season"],
        "Scenario": row["Scenario"],
        "Demand change (MW)": row["Mean Change (MW)"],
        "Energy change (GWh/season)": row["Mean Change (MW)"] * hours / 1000,
        "CO2 (t/season)": mw_to_tonnes(row["Mean Change (MW)"], hours, MARGINAL_FACTOR),
        "CO2 low (t)":  mw_to_tonnes(row["95% CI lower"], hours, MARGINAL_FACTOR),
        "CO2 high (t)": mw_to_tonnes(row["95% CI upper"], hours, MARGINAL_FACTOR),
        "Classification": row["Classification"],
    })

co2_df = pd.DataFrame(co2_rows)
print(co2_df.round(1).to_string(index=False))

print("\n" + "=" * 66)
print("ANNUAL NET EFFECT (test year)")
print("=" * 66)
for scen in scen_cols:
    sub = co2_df[co2_df["Scenario"] == scen]
    net = sub["CO2 (t/season)"].sum()
    up  = sub.loc[sub["CO2 (t/season)"] > 0, "CO2 (t/season)"].sum()
    dn  = sub.loc[sub["CO2 (t/season)"] < 0, "CO2 (t/season)"].sum()
    print(f"\n{scen}")
    print(f"  Seasons with higher emissions: {up:+12,.0f} t CO2")
    print(f"  Seasons with lower emissions:  {dn:+12,.0f} t CO2")
    print(f"  NET:                           {net:+12,.0f} t CO2")

print("""
IMPORTANT — how to report this.

The net annual figure inherits the annual confidence interval from Cell 8.8,
which CONTAINS ZERO. Do not present the net as an established increase.
The defensible statement is:

  "Under +3 °C, the model predicts a seasonal redistribution of emissions —
   an increase of X tonnes in summer and a decrease of Y tonnes in winter —
   with a net annual change that is not statistically distinguishable from
   zero at the 95% level."

The seasonal figures are your result. The net is a null result, and reporting
it as such is a strength.
""")
```

```python
# ============================================================
# PHASE 10 — CELL 4 — How much does the factor choice matter?
# ============================================================

factor_options = {
    f"Marginal (from data, {MARGINAL_FACTOR:.3f})": MARGINAL_FACTOR,
    f"Average (from data, {implied_average:.3f})":  implied_average,
    "National average (0.321)":                     0.321,
    "Gas-only upper bound (0.49)":                   0.49,   # VERIFY & CITE
}

print("SENSITIVITY OF THE +3 °C RESULT TO THE EMISSION FACTOR\n")

sens_rows = []
sub3 = co2_df[co2_df["Scenario"] == "+3°C"]
for label, factor in factor_options.items():
    summer = mw_to_tonnes(
        sub3.loc[sub3["Season"] == "summer", "Demand change (MW)"].iloc[0],
        season_hours["summer"], factor)
    net = sum(mw_to_tonnes(r["Demand change (MW)"], season_hours[r["Season"]], factor)
              for _, r in sub3.iterrows())
    sens_rows.append({"Factor": label, "kg CO2/kWh": factor,
                      "Summer (t)": summer, "Net annual (t)": net})

sens_df = pd.DataFrame(sens_rows)
print(sens_df.round(1).to_string(index=False))

spread = sens_df["Net annual (t)"].max() - sens_df["Net annual (t)"].min()
print(f"\nRange across factor choices: {spread:,.0f} tonnes")
print("If this range is wide relative to the estimates themselves, the CO2")
print("result is more sensitive to your emission-factor assumption than to")
print("the machine learning. Say so plainly in Limitations.")

fig, ax = plt.subplots(figsize=(11, 5.5))
x = np.arange(len(season_order))
width = 0.25
for i, scen in enumerate(scen_cols):
    sub = co2_df[co2_df["Scenario"] == scen].set_index("Season").reindex(season_order)
    lower_err = sub["CO2 (t/season)"] - sub["CO2 low (t)"]
    upper_err = sub["CO2 high (t)"] - sub["CO2 (t/season)"]
    ax.bar(x + (i - 1) * width, sub["CO2 (t/season)"], width, label=scen,
           yerr=[lower_err.abs().values, upper_err.abs().values], capsize=3,
           error_kw={"linewidth": 0.9})
ax.axhline(0, color="black", linewidth=0.8)
ax.set_xticks(x)
ax.set_xticklabels(season_order)
ax.set_xlabel("Season")
ax.set_ylabel("Estimated Change in CO₂ Emissions (tonnes per season)")
ax.set_title(f"Estimated Seasonal CO₂ Impact "
             f"(marginal factor {MARGINAL_FACTOR:.3f} kg/kWh, 95% CI)")
ax.legend(title="Scenario")
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
